Para el manual de bienvenida de los nuevos: el organigrama comercial dibujado como un árbol de texto,
a partir del fichero de RRHH. Sin base de datos.

```python
import csv

from pathlib import Path

RAIZ = Path.home() / "master-data-ia"
ruta = RAIZ / "datos" / "rrhh" / "comerciales.csv"
print("Ruta:", ruta, "| existe:", ruta.exists())

with open(ruta, encoding="latin-1", newline="") as f:
    personas = list(csv.DictReader(f))

nombre = {p["comercial_id"]: f"{p['nombre']} {p['apellidos']}" for p in personas}
puesto = {p["comercial_id"]: p["puesto"] for p in personas}
de_baja = {p["comercial_id"] for p in personas if p["fecha_baja"]}

subordinados = {}
for p in personas:
    subordinados.setdefault(p["jefe_id"], []).append(p["comercial_id"])

def dibujar(persona_id, nivel=0):
    marca = " (de baja)" if persona_id in de_baja else ""
    print(" " * nivel + f"{nombre[persona_id]} · {puesto[persona_id]}{marca}")
    for sub in subordinados.get(persona_id, []):
        dibujar(sub, nivel + 1)

for raiz in subordinados.get("", []):
    dibujar(raiz)
```
