Me han dicho que lo del MJD-102 (referencias por almacén) salió con SKU mal escritos. ¿Me lo repites
bien? Y esta vez sin base de datos: el fichero de almacén tal cual llega, porque quiero poder hacerlo yo
cada vez que lo exporte.

- Leer directamente datos/almacen/inventario_2026-09-30.csv.
- Referencias distintas (con el SKU normalizado) y unidades totales por almacén.
- Cuántas filas tenían el SKU mal escrito.

```python
import csv
from pathlib import Path
import marejada_utils as mu

RAIZ = Path.home() / "master-data-ia"
ruta = RAIZ / "datos" / "almacen" / "inventario_2026-09-30.csv"
print("Ruta:", ruta, "| existe:", ruta.exists())

referencias = {}   # almacén -> conjunto de SKU normalizados
unidades = {}      # almacén -> unidades totales
mal_escritos = 0

with open(ruta, encoding="utf-8", newline="") as f:
    for fila in csv.DictReader(f):
        almacen = fila["almacen"]
        sku = mu.normalizar_sku(fila["sku"])
        if sku != fila["sku"]:
            mal_escritos += 1
        referencias.setdefault(almacen, set()).add(sku)
        unidades[almacen] = unidades.get(almacen, 0) + int(fila["unidades"])

for almacen in sorted(referencias):
    print(f"{almacen:<10} {len(referencias[almacen]):>4} referencias {unidades[almacen]:>7,} unidades")
print(f"Filas con el SKU mal escrito: {mal_escritos}")
```
